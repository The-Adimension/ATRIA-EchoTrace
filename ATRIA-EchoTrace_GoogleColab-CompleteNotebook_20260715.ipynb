````python
#@title Configure Data Paths, Constants, and Sample Size

# ===== Echocardiographic_frame Pre-processed Data =====
# Generated by preprocess_echocardiographic_frame.py:
#   echocardiographic_frame_processed/frames/      → 2000 PNGs  (patient{NNNN}_{view}_{instant}.png)
#   echocardiographic_frame_processed/tracings.json → polygon coords per image (LV + LA, 30 pts each)
#   echocardiographic_frame_processed/metadata.csv  → patient-level clinical data + split info

dataset_root = "/content/echocardiographic_frame_processed/camus_processed/"  # @param {type: "string"}

# ===== TARGET STRUCTURE =====
# Switch between "LV" and "LA" to train on one structure at a time.
# "LV" → left ventricle endocardial tracing
# "LA" → left atrium tracing
TARGET_STRUCTURE = "LV"  # @param ["LV", "LA"]

# Derived label / description used in prompts and evaluation
STRUCTURE_INFO = {
    "LV": {"label": "left_ventricle_endocardium", "name": "left ventricle endocardial border",
            "short": "LV endocardium", "poly_key": "lv_polygon", "has_key": "has_lv"},
    "LA": {"label": "left_atrium",                "name": "left atrium border",
            "short": "LA",             "poly_key": "la_polygon", "has_key": "has_la"},
}
STRUCT = STRUCTURE_INFO[TARGET_STRUCTURE]  # convenience alias

# Derived paths
FRAMES_DIR    = os.path.join(dataset_root, "frames")
TRACINGS_PATH = os.path.join(dataset_root, "tracings.json")
METADATA_PATH = os.path.join(dataset_root, "metadata.csv")

# Coordinate normalisation (same as EchoNet pipeline)
NORM_SCALE = 1000  # Polygon coords normalised to [0, 1000]

# SAMPLE SIZE CONTROLS — set to None to use all available
TRAIN_MAX_SAMPLES = 1000   # @param {type: "integer"} - Max training samples (None = all ~1600)
VAL_MAX_SAMPLES   = 200   # @param {type: "integer"} - Max validation samples (None = all ~200)
TEST_MAX_SAMPLES  = 200   # @param {type: "integer"} - Max test samples (None = all ~200)

# Model configuration
MODEL_ID = "google/medgemma-1.5-4b-it"  # @param {type: "string"}

# Verify paths
print("=" * 60)
print(f"ECHOCARDIOGRAPHIC_FRAME DATA PATHS  —  TARGET: {TARGET_STRUCTURE} ({STRUCT['short']})")
print("=" * 60)
print(f"Echocardiographic_frame root:         {dataset_root}")
print(f"Frames dir exists:  {os.path.exists(FRAMES_DIR)}")
print(f"Tracings exists:    {os.path.exists(TRACINGS_PATH)}")
print(f"Metadata exists:    {os.path.exists(METADATA_PATH)}")

if os.path.exists(FRAMES_DIR):
    sample_files = sorted(os.listdir(FRAMES_DIR))[:5]
    print(f"Sample PNGs: {sample_files}")
    print(f"Total PNGs:  {len(os.listdir(FRAMES_DIR))}")

print("\n" + "=" * 60)
print("SAMPLE SIZE CONFIGURATION")
print("=" * 60)
print(f"Target structure:   {TARGET_STRUCTURE}  →  {STRUCT['name']}")
print(f"Training samples:   {TRAIN_MAX_SAMPLES if TRAIN_MAX_SAMPLES else 'ALL'}")
print(f"Validation samples: {VAL_MAX_SAMPLES if VAL_MAX_SAMPLES else 'ALL'}")
print(f"Test samples:       {TEST_MAX_SAMPLES if TEST_MAX_SAMPLES else 'ALL'}")
print(f"\nModel: {MODEL_ID}")
````

````python
#@title Load Echocardiographic frame Tracings and Metadata

def load_echocardiographic_frame_data(tracings_path: str, metadata_path: str):
    """
    Load pre-processed Echocardiographic_frame data produced by preprocess_echocardiographic_frame.py.

    Returns:
        tracings_dict : dict[str, dict]  — keyed by image key (e.g. 'patient0001_4CH_ED')
        metadata_df   : pd.DataFrame     — one row per image with clinical info + split
    """
    with open(tracings_path, "r") as f:
        tracings_dict = json.load(f)

    metadata_df = pd.read_csv(metadata_path)

    return tracings_dict, metadata_df


# Load data
tracings_dict, metadata_df = load_echocardiographic_frame_data(TRACINGS_PATH, METADATA_PATH)

print(f"Loaded {len(tracings_dict)} tracings from tracings.json")
print(f"Loaded {len(metadata_df)} rows from metadata.csv")
print(f"\nSplit distribution:")
print(metadata_df["split"].value_counts().to_string())
print(f"\nView distribution:")
print(metadata_df["view"].value_counts().to_string())
print(f"\nInstant distribution:")
print(metadata_df["instant"].value_counts().to_string())
````

````python
#@title Visualise pre-computed polygon for the target structure from Echocardiograhic frame ground-truth segmentation masks

def load_frame(key: str, frames_dir: str) -> Image.Image:
    """
    Load a pre-extracted Echocardiographic_frame frame as RGB PIL Image.

    Args:
        key: Image key e.g. 'patient0001_4CH_ED'
        frames_dir: Path to echocardiographic_frame_processed/frames/
    Returns:
        PIL Image in RGB mode
    """
    img_path = os.path.join(frames_dir, f"{key}.png")
    if not os.path.exists(img_path):
        raise FileNotFoundError(f"Frame not found: {img_path}")
    img = Image.open(img_path)
    return img.convert("RGB") if img.mode != "RGB" else img


def visualize_polygon_on_image(image: Image.Image, polygon: list, title: str = ""):
    """
    Visualise a single polygon overlay on an Echocardiographic_frame frame.
    Polygon is in [[y, x], ...] format normalised to [0, 1000].
    """
    img_w, img_h = image.size

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    # 1. Original image
    axes[0].imshow(image)
    axes[0].set_title("Original Frame")
    axes[0].axis("off")

    # 2. Polygon overlay
    overlay_img = image.copy().convert("RGBA")
    overlay = Image.new("RGBA", overlay_img.size, (0, 0, 0, 0))
    draw = ImageDraw.Draw(overlay)

    px = [(p[1] / NORM_SCALE * img_w, p[0] / NORM_SCALE * img_h) for p in polygon]
    if len(px) >= 3:
        draw.polygon(px, outline=(0, 255, 0, 255), fill=(0, 255, 0, 60))
        for x, y in px:
            draw.ellipse([x - 2, y - 2, x + 2, y + 2], fill=(255, 0, 0, 255))

    axes[1].imshow(Image.alpha_composite(overlay_img, overlay))
    axes[1].set_title(f"{STRUCT['short']} ({len(polygon)} pts)")
    axes[1].axis("off")

    plt.suptitle(title, fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.show()

    print(f"{TARGET_STRUCTURE} polygon: {len(polygon)} points")


# ─── TEST: Visualise one example per view ───
poly_key = STRUCT["poly_key"]

print("=" * 60)
print(f"EXAMPLE: Echocardiographic_frame Frames with {TARGET_STRUCTURE} Polygon")
print("=" * 60)

for view_instant in ["4CH_ED", "4CH_ES", "2CH_ED", "2CH_ES"]:
    test_key = None
    for k in tracings_dict:
        if view_instant in k:
            test_key = k
            break

    if test_key is None:
        print(f"No sample found for {view_instant}")
        continue

    entry = tracings_dict[test_key]
    polygon = entry.get(poly_key)
    if not polygon:
        print(f"No {TARGET_STRUCTURE} polygon for {test_key}")
        continue

    test_image = load_frame(test_key, FRAMES_DIR)

    print(f"\n--- {test_key} ({test_image.size[0]}x{test_image.size[1]}) ---")
    raw_key = TARGET_STRUCTURE.lower() + "_points_raw"
    raw_pts = entry.get(raw_key, "?")
    visualize_polygon_on_image(
        test_image, polygon,
        title=f"{test_key}  ({TARGET_STRUCTURE}, raw pts: {raw_pts} → 30)",
    )
````

````python
#@title Load pre-extracted echocardiographic frames: In PNG format, can be variable resolution- load_frame defined in previous cell.

# Quick sanity check — load one frame per split
for split_name in ["train", "val", "test"]:
    split_keys = [k for k, v in tracings_dict.items() if v["split"] == split_name]
    if not split_keys:
        print(f"No keys for split '{split_name}'")
        continue

    test_key = split_keys[0]
    try:
        img = load_frame(test_key, FRAMES_DIR)
        entry = tracings_dict[test_key]
        print(f"[{split_name:5s}] {test_key}: loaded {img.size} (mode={img.mode}), "
              f"stored size=({entry['image_w']}x{entry['image_h']}), "
              f"LV={len(entry['lv_polygon'])} pts, "
              f"LA={'None' if entry.get('la_polygon') is None else str(len(entry['la_polygon'])) + ' pts'}")
    except FileNotFoundError as e:
        print(f"[{split_name:5s}] ERROR: {e}")
````

````python
#@title Define Prompt Template (single-structure, config-driven)

# View-name mapping for natural-language prompts
VIEW_NAMES = {"2CH": "2-chamber", "4CH": "4-chamber"}
INSTANT_NAMES = {"ED": "end-diastole", "ES": "end-systole"}

# Prompt template: one structure at a time, and use double braces {{ }} to escape literal JSON keys for .format()
PROMPT_TEMPLATE = (
    "Instructions:\n"
    "The following user query will require outputting polygon coordinates for "
    "{structure_name} tracing. "
    "The format of polygon coordinates is [[y0, x0], [y1, x1], ...] where each point is "
    "[y, x] representing the boundary. Always normalize the x and y coordinates to the "
    "range [0, 1000], meaning that a point at 15% of the image width would be associated "
    "with an x coordinate of 150. You MUST output a single parseable json list of objects "
    'enclosed into ```json...``` brackets, for instance ```json[{{"polygon_2d": '
    '[[100, 200], [150, 250]], "label": "{structure_label}"}}]``` is a valid output. '
    "Now answer to the user query.\n\n"
    "Query:\n"
    "This is an apical {view_name} echocardiogram view at {instant_name}. "
    "Trace the {structure_name}. "
    'Output the final answer in the format "Final Answer: X" where X is a JSON list of '
    'objects with "polygon_2d" and "label" keys. Answer:'
)

def create_ground_truth_response(polygon: list) -> str:
    """Create the expected model response with a single JSON polygon."""
    objects = [{"polygon_2d": polygon, "label": STRUCT["label"]}]
    gt_json = json.dumps(objects)
    return (
        f"The {STRUCT['name']} traces the inner wall of the heart chamber.\n\n"
        f"Final Answer: ```json{gt_json}```"
    )

# Test prompt and response
test_key = list(tracings_dict.keys())[0]
test_entry = tracings_dict[test_key]

test_prompt = PROMPT_TEMPLATE.format(
    structure_name=STRUCT["name"],
    structure_label=STRUCT["label"],
    view_name=VIEW_NAMES[test_entry["view"]],
    instant_name=INSTANT_NAMES[test_entry["instant"]],
)
test_response = create_ground_truth_response(test_entry[STRUCT["poly_key"]])

print("=" * 60)
print(f"TARGET STRUCTURE: {TARGET_STRUCTURE} ({STRUCT['short']})")
print(f"PROMPT (for {test_key}):")
print("=" * 60)
print(test_prompt)
print("\n" + "=" * 60)
print("EXPECTED RESPONSE (truncated):")
print("=" * 60)
print(test_response + "..." if len(test_response) > 500 else test_response)
````

````python
#@title  Prepare Dataset Samples —  (single-structure)

def prepare_echocardiographic_frame_samples(
    tracings_dict: dict,
    metadata_df: pd.DataFrame,
    frames_dir: str,
    split: str = "train",
    max_samples: int = None,
) -> list:
    """
    Prepare Echocardiographic_frame dataset samples for fine-tuning.
    Only includes samples where TARGET_STRUCTURE polygon exists.
    """
    poly_key = STRUCT["poly_key"]
    split_df = metadata_df[metadata_df["split"] == split]
    keys_in_split = split_df["key"].tolist()

    samples = []
    skipped_no_png = 0
    skipped_no_poly = 0

    for key in keys_in_split:
        if max_samples and len(samples) >= max_samples:
            break

        entry = tracings_dict.get(key)
        if entry is None:
            continue

        polygon = entry.get(poly_key)
        if not polygon or len(polygon) < 3:
            skipped_no_poly += 1
            continue

        png_path = os.path.join(frames_dir, f"{key}.png")
        if not os.path.exists(png_path):
            skipped_no_png += 1
            continue

        samples.append({
            "key": key,
            "frames_dir": frames_dir,
            "view": entry["view"],
            "instant": entry["instant"],
            "polygon": polygon,           # single target structure
            "image_h": entry["image_h"],
            "image_w": entry["image_w"],
            "patient_id": entry["patient_id"],
        })

    print(f"Split '{split}': {len(samples)} samples prepared  (target={TARGET_STRUCTURE})")
    print(f"  - Skipped (no {TARGET_STRUCTURE}):  {skipped_no_poly}")
    print(f"  - Skipped (no PNG): {skipped_no_png}")
    return samples


# Prepare samples
print("=" * 60)
print(f"PREPARING ECHOCARDIOGRAPHIC_FRAME DATASET SAMPLES  —  {TARGET_STRUCTURE}")
print("=" * 60)

train_samples = prepare_echocardiographic_frame_samples(
    tracings_dict, metadata_df, FRAMES_DIR,
    split="train", max_samples=TRAIN_MAX_SAMPLES,
)
val_samples = prepare_echocardiographic_frame_samples(
    tracings_dict, metadata_df, FRAMES_DIR,
    split="val", max_samples=VAL_MAX_SAMPLES,
)
test_samples = prepare_echocardiographic_frame_samples(
    tracings_dict, metadata_df, FRAMES_DIR,
    split="test", max_samples=TEST_MAX_SAMPLES,
)

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Target structure:   {TARGET_STRUCTURE} ({STRUCT['short']})")
print(f"Training samples:   {len(train_samples)}")
print(f"Validation samples: {len(val_samples)}")
print(f"Test samples:       {len(test_samples)}")
print(f"Total:              {len(train_samples) + len(val_samples) + len(test_samples)}")
````

````python
#@title Define Custom Data Collator

def collate_fn(examples: list[dict[str, Any]]):
    """
    Custom data collator for multimodal LV tracing task.
    Handles images and text, creates labels with proper masking.
    """
    texts = []
    images = []

    for example in examples:
        # Convert image to RGB if needed
        img = example["image"]
        if img.mode != "RGB":
            img = img.convert("RGB")
        images.append([img])

        # Apply chat template
        text = processor.apply_chat_template(
            example["messages"],
            add_generation_prompt=False,
            tokenize=False
        ).strip()
        texts.append(text)

    # Tokenize and process images
    batch = processor(
        text=texts,
        images=images,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=2048
    )

    # Create labels (same as input_ids but with masking)
    labels = batch["input_ids"].clone()

    # Get special token IDs for masking
    pad_token_id = processor.tokenizer.pad_token_id

    # Get image token ID (handle different tokenizer versions)
    try:
        image_token_id = processor.tokenizer.convert_tokens_to_ids(
            processor.tokenizer.special_tokens_map.get("boi_token", "<image>")
        )
    except:
        image_token_id = None

    # Mask tokens not used in loss computation
    labels[labels == pad_token_id] = -100
    if image_token_id is not None:
        labels[labels == image_token_id] = -100
    labels[labels == 262144] = -100  # Additional image token ID

    batch["labels"] = labels
    return batch


# Test the collator
print("Testing data collator...")
test_batch = collate_fn([train_data[0]])
print(f"Batch keys: {list(test_batch.keys())}")
print(f"Input IDs shape: {test_batch['input_ids'].shape}")
print(f"Labels shape: {test_batch['labels'].shape}")
print(f"Pixel values shape: {test_batch['pixel_values'].shape}")
````

````python
#@title Evaluation Utilities

import skimage.draw


def parse_polygon_from_response(response: str) -> Optional[list]:
    """
    Extract the target structure's polygon from model response.
    Returns a list of [y, x] points or None.
    """
    # Remove thinking trace if present
    if "<unused95>" in response:
        response = response.split("<unused95>", 1)[1].lstrip()

    json_match = re.search(r'```json\s*(.*?)\s*```', response, re.DOTALL)
    if json_match:
        try:
            data = json.loads(json_match.group(1))
            if isinstance(data, list):
                for obj in data:
                    label = obj.get("label", "").lower()
                    poly = obj.get("polygon_2d")
                    # Match whatever the current target structure is
                    target_label = STRUCT["label"].lower()
                    if target_label.replace("_", " ") in label.replace("_", " "):
                        return poly
                    # Fallback: if only one object, assume it's the target
                if len(data) == 1 and "polygon_2d" in data[0]:
                    return data[0]["polygon_2d"]
        except json.JSONDecodeError:
            pass
    return None


def polygon_to_mask(polygon: list, image_h: int, image_w: int) -> np.ndarray:
    """
    Convert normalised [0,1000] polygon to binary mask at the image's native resolution.
    """
    if not polygon or len(polygon) < 3:
        return np.zeros((image_h, image_w), dtype=np.float32)

    x_coords = np.array([p[1] / NORM_SCALE * image_w for p in polygon])
    y_coords = np.array([p[0] / NORM_SCALE * image_h for p in polygon])

    rr, cc = skimage.draw.polygon(
        y_coords.astype(int),
        x_coords.astype(int),
        (image_h, image_w),
    )

    mask = np.zeros((image_h, image_w), dtype=np.float32)
    mask[rr, cc] = 1
    return mask


def compute_dice(pred_mask: np.ndarray, gt_mask: np.ndarray) -> float:
    """Compute Dice coefficient between masks."""
    intersection = np.sum(pred_mask * gt_mask)
    union = np.sum(pred_mask) + np.sum(gt_mask)
    if union == 0:
        return 1.0 if np.sum(pred_mask) == 0 else 0.0
    return 2.0 * intersection / union


def compute_iou(pred_mask: np.ndarray, gt_mask: np.ndarray) -> float:
    """Compute IoU between masks."""
    intersection = np.sum(pred_mask * gt_mask)
    union = np.sum(np.logical_or(pred_mask, gt_mask))
    return intersection / union if union > 0 else 1.0


print(f"Evaluation utilities loaded — target: {TARGET_STRUCTURE} ({STRUCT['short']})")
````

````python
# @title
import os
import io
import re
import json
import base64
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw
import torch
from transformers import AutoProcessor, AutoModelForImageTextToText, BitsAndBytesConfig
from peft import PeftModel
from IPython.display import display, HTML, Javascript, JSON
from google.colab import output
import gc

# --- Constants & Helpers ---
NORM_SCALE = 1000
TARGET_STRUCTURE = "LV"
STRUCTURE_INFO = {
    "LV": {"label": "left_ventricle_endocardium", "name": "left ventricle endocardial border",
            "short": "LV endocardium", "poly_key": "lv_polygon", "has_key": "has_lv"},
}
STRUCT = STRUCTURE_INFO[TARGET_STRUCTURE]
VIEW_NAMES = {"2CH": "2-chamber", "4CH": "4-chamber"}
INSTANT_NAMES = {"ED": "end-diastole", "ES": "end-systole"}

# FIXED: Use double braces {{ }} for literal JSON keys to avoid KeyError during .format()
PROMPT_TEMPLATE = (
    "Instructions:\n"
    "The following user query will require outputting polygon coordinates for "
    "{structure_name} tracing. "
    "The format of polygon coordinates is [[y0, x0], [y1, x1], ...] where each point is "
    "[y, x] representing the boundary. Always normalize the x and y coordinates to the "
    "range [0, 1000], meaning that a point at 15% of the image width would be associated "
    "with an x coordinate of 150. You MUST output a single parseable json list of objects "
    'enclosed into ```json...``` brackets, for instance ```json[{{"polygon_2d": '
    '[[100, 200], [150, 250]], "label": "{structure_label}"}}]``` is a valid output. '
    "Now answer to the user query.\n\n"
    "Query:\n"
    "This is an apical echocardiogram view. Trace the {structure_name}. "
    'Output the final answer in the format "Final Answer: X" where X is a JSON list of '
    'objects with "polygon_2d" and "label" keys. Answer:'
)

def parse_polygon(response: str) -> list | None:
    if "<unused95>" in response:
        response = response.split("<unused95>", 1)[1].lstrip()
    json_match = re.search(r'```json\s*(.*?)\s*```', response, re.DOTALL)
    if json_match:
        try:
            data = json.loads(json_match.group(1))
            if isinstance(data, list):
                for obj in data:
                    label = obj.get("label", "").lower()
                    if STRUCT["label"].lower().replace("_", " ") in label.replace("_", " "):
                        return obj.get("polygon_2d")
                if len(data) == 1 and "polygon_2d" in data[0]:
                    return data[0]["polygon_2d"]
        except:
            pass
    return None

def draw_polygon_on_image(img, poly, color, NORM_SCALE=1000):
    if not poly or len(poly) < 3:
        return img.convert("RGB")
    overlay = Image.new("RGBA", img.size, (0, 0, 0, 0))
    draw = ImageDraw.Draw(overlay)
    w, h = img.size
    pts = [(p[1] / NORM_SCALE * w, p[0] / NORM_SCALE * h) for p in poly]
    draw.polygon(pts, outline=color, fill=color[:3] + (60,))
    return Image.alpha_composite(img.convert("RGBA"), overlay).convert("RGB")

# --- Model Global State ---
ft_model = None
ft_processor = None
raw_images = [None, None]
model_poly = [[], []] # Store initial model predictions
user_poly = [[], []]  # Store user revisions

def load_model(lora_path: str):
    global ft_model, ft_processor
    if ft_model is not None:
        return

    if 'model' in globals() and 'processor' in globals():
        print("Reusing pre-loaded model from notebook namespace...")
        ft_model = globals()['model']
        ft_processor = globals()['processor']
        return

    print("Loading base model and LoRA adapter...")
    gc.collect(); torch.cuda.empty_cache()
    MODEL_ID = "google/medgemma-1.5-4b-it"
    processor = AutoProcessor.from_pretrained(MODEL_ID)
    processor.tokenizer.padding_side = "left"
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True, bnb_4bit_use_double_quant=True, bnb_4bit_quant_type="nf4", bnb_4bit_compute_dtype=torch.bfloat16,
    )
    base_model = AutoModelForImageTextToText.from_pretrained(
        MODEL_ID, quantization_config=bnb_config, torch_dtype=torch.bfloat16, device_map="auto", attn_implementation="eager"
    )
    ft_model = PeftModel.from_pretrained(base_model, lora_path)
    ft_model.eval()
    ft_processor = processor
    print("Model loaded!")

# --- Backend Callbacks ---
def process_image_backend(b64_img, lora_path, img_idx):
    global raw_images, model_poly
    try:
        load_model(lora_path)
        img_data = base64.b64decode(b64_img.split(',')[1])
        image = Image.open(io.BytesIO(img_data)).convert("RGB")
        raw_images[img_idx] = image

        instant = "ED" if img_idx == 0 else "ES"
        prompt_text = PROMPT_TEMPLATE.format(
            structure_name=STRUCT["name"], structure_label=STRUCT["label"],
            view_name=VIEW_NAMES["2CH"], instant_name=INSTANT_NAMES[instant]
        )
        messages = [{"role": "user", "content": [{"type": "image", "image": image}, {"type": "text", "text": prompt_text}]}]

        gc.collect(); torch.cuda.empty_cache()
        text = ft_processor.apply_chat_template(messages, add_generation_prompt=True, tokenize=False)
        inputs = ft_processor(text=text, images=[image], return_tensors="pt").to(ft_model.device)

        with torch.inference_mode():
            output_ids = ft_model.generate(**inputs, max_new_tokens=1024, do_sample=False, pad_token_id=ft_processor.tokenizer.eos_token_id)

        input_len = inputs["input_ids"].shape[1]
        response = ft_processor.decode(output_ids[0][input_len:], skip_special_tokens=True)
        poly = parse_polygon(response) or []
        model_poly[img_idx] = poly

        return JSON({"status": "success", "polygon": poly, "idx": img_idx})
    except Exception as e:
        import traceback
        traceback.print_exc()
        return JSON({"status": "error", "message": str(e)})

def save_polygon_backend(poly1, poly2):
    global user_poly, raw_images, model_poly
    user_poly[0] = poly1
    user_poly[1] = poly2

    output_dir = "/content/tracing_results"
    os.makedirs(output_dir, exist_ok=True)
    saved_logs = []

    for i in range(2):
        img = raw_images[i]
        if img is None:
            continue

        m_poly = model_poly[i]
        u_poly = user_poly[i]

        # 1. Save Coordinates to JSON
        json_data = {
            "image_index": i,
            "model_polygon_2d": m_poly,
            "user_polygon_2d": u_poly
        }
        json_path = os.path.join(output_dir, f"tracing_data_img{i}.json")
        with open(json_path, "w") as f:
            json.dump(json_data, f, indent=4)

        # 2. Generate side-by-side Visualizations
        fig, axes = plt.subplots(1, 4, figsize=(20, 5))

        axes[0].imshow(img)
        axes[0].set_title(f"Original Image {i+1}")
        axes[0].axis("off")

        img_model = draw_polygon_on_image(img, m_poly, (255, 0, 0, 255)) # Red
        axes[1].imshow(img_model)
        axes[1].set_title(f"Model Prediction ({len(m_poly)} pts)")
        axes[1].axis("off")

        img_user = draw_polygon_on_image(img, u_poly, (0, 255, 0, 255)) # Green
        axes[2].imshow(img_user)
        axes[2].set_title(f"User Revision ({len(u_poly)} pts)")
        axes[2].axis("off")

        img_overlay = draw_polygon_on_image(img_model, u_poly, (0, 255, 0, 255))
        axes[3].imshow(img_overlay)
        axes[3].set_title("Overlay (Red=Model, Green=User)")
        axes[3].axis("off")

        plt.tight_layout()
        plot_path = os.path.join(output_dir, f"tracing_vis_img{i}.png")
        plt.savefig(plot_path, dpi=150)
        plt.close(fig)

        saved_logs.append(f"Img {i+1} saved: JSON and PNG to {output_dir}")

    print(f"Saved user revised polygons: Image1={len(poly1)} points, Image2={len(poly2)} points.")
    return JSON({"status": "saved", "logs": saved_logs})

output.register_callback('process_image_backend', process_image_backend)
output.register_callback('save_polygon_backend', save_polygon_backend)

# --- HTML/JS UI ---
html_code = """
<style>
  .main-container { display: flex; flex-direction: column; gap: 10px; font-family: sans-serif; }
  .top-controls { display: flex; gap: 10px; align-items: center; background: #e9ecef; padding: 10px; border-radius: 5px; }
  .editors-container { display: flex; gap: 20px; width: 100%; }
  .editor-pane { flex: 1; display: flex; flex-direction: column; gap: 10px; border: 1px solid #ccc; padding: 10px; border-radius: 5px; background: #fdfdfd; }
  canvas { border: 1px solid #999; cursor: crosshair; max-width: 100%; background: #eee; }
  #logs { height: 100px; overflow-y: auto; background: #fff; padding: 5px; font-size: 12px; border: 1px solid #ccc; }
  button, input { padding: 5px; }
</style>

<div class="main-container">
  <h3>Interactive Dual MedGemma Echocardiographic_frame Editor</h3>

  <div class="top-controls">
    <label>LoRA Path:</label>
    <input type="text" id="loraPath" value="/content/drive/MyDrive/ATRIA-G3 (1)/output_20K/checkpoint-1034" style="width: 300px;">
    <button id="saveBtn" style="background: #28a745; color: white; border: none; font-weight: bold; cursor: pointer;">Save Both Polygons & Generate Docs</button>
  </div>

  <div id="logs">Waiting for action...</div>

  <div class="editors-container">
    <!-- Image 1 (ED) -->
    <div class="editor-pane">
      <h4>Image 1 (e.g. End-Diastole)</h4>
      <div>
        <input type="file" id="fileUpload0" accept="image/*">
        <button id="processBtn0">Process Image 1</button>
      </div>
      <canvas id="editorCanvas0"></canvas>
    </div>

    <!-- Image 2 (ES) -->
    <div class="editor-pane">
      <h4>Image 2 (e.g. End-Systole)</h4>
      <div>
        <input type="file" id="fileUpload1" accept="image/*">
        <button id="processBtn1">Process Image 2</button>
      </div>
      <canvas id="editorCanvas1"></canvas>
    </div>
  </div>
</div>

<script>
  const loraPath = document.getElementById('loraPath');
  const logs = document.getElementById('logs');
  const saveBtn = document.getElementById('saveBtn');

  function log(msg) { logs.innerHTML += `<div>${msg}</div>`; logs.scrollTop = logs.scrollHeight; }

  // State for both editors
  const editors = [
    { id: 0, canvas: document.getElementById('editorCanvas0'), fileInput: document.getElementById('fileUpload0'), processBtn: document.getElementById('processBtn0'), img: new Image(), points: [], draggingPointIndex: -1, currentBase64: null },
    { id: 1, canvas: document.getElementById('editorCanvas1'), fileInput: document.getElementById('fileUpload1'), processBtn: document.getElementById('processBtn1'), img: new Image(), points: [], draggingPointIndex: -1, currentBase64: null }
  ];

  editors.forEach(ed => {
    ed.ctx = ed.canvas.getContext('2d');
    const pointRadius = 5;

    // Draw function
    ed.draw = () => {
      if (!ed.img.src) return;
      ed.canvas.width = ed.img.width;
      ed.canvas.height = ed.img.height;
      ed.ctx.clearRect(0, 0, ed.canvas.width, ed.canvas.height);
      ed.ctx.drawImage(ed.img, 0, 0);

      if (ed.points.length > 0) {
        ed.ctx.beginPath();
        ed.ctx.moveTo(ed.points[0].x, ed.points[0].y);
        for (let i = 1; i < ed.points.length; i++) ed.ctx.lineTo(ed.points[i].x, ed.points[i].y);
        ed.ctx.lineTo(ed.points[0].x, ed.points[0].y);
        ed.ctx.fillStyle = 'rgba(255, 0, 0, 0.2)';
        ed.ctx.fill();
        ed.ctx.strokeStyle = 'red';
        ed.ctx.lineWidth = 2;
        ed.ctx.stroke();

        ed.points.forEach((p, i) => {
          ed.ctx.beginPath();
          ed.ctx.arc(p.x, p.y, pointRadius, 0, Math.PI * 2);
          ed.ctx.fillStyle = i === ed.draggingPointIndex ? 'yellow' : 'red';
          ed.ctx.fill();
          ed.ctx.stroke();
        });
      }
    };

    ed.img.onload = ed.draw;

    // Handle upload
    ed.fileInput.addEventListener('change', (e) => {
      const file = e.target.files[0];
      if (!file) return;
      const reader = new FileReader();
      reader.onload = (ev) => {
        ed.currentBase64 = ev.target.result;
        ed.img.src = ed.currentBase64;
        ed.points = [];
        log(`Image ${ed.id + 1} loaded.`);
      };
      reader.readAsDataURL(file);
    });

    // Handle Process
    ed.processBtn.addEventListener('click', async () => {
      if (!ed.currentBase64) return log(`Please upload Image ${ed.id + 1} first.`);
      log(`Processing Image ${ed.id + 1}...`);
      try {
        const res = await google.colab.kernel.invokeFunction('process_image_backend', [ed.currentBase64, loraPath.value, ed.id], {});
        const data = res.data && res.data['application/json'];

        if (!data) {
           log(`Error on Image ${ed.id + 1}: Unexpected empty response from backend.`);
           return;
        }
        if (data.status === "error") {
           log(`Backend Error on Image ${ed.id + 1}: ${data.message}`);
           return;
        }
        if (data.polygon && data.polygon.length > 0) {
          ed.points = data.polygon.map(p => (
            {
            x: (p[1] / 1000) * ed.img.width,
            y: (p[0] / 1000) * ed.img.height
          }));
          log(`Successfully received ${ed.points.length} points for Image ${ed.id + 1}.`);
          ed.draw();
        } else {
          log(`Failed to parse polygon for Image ${ed.id + 1}. The model may have returned an empty trace.`);
        }
      } catch (err) {
        log(`Error invoking backend for Image ${ed.id + 1}: ` + err.message);
      }
    });

    // Interaction
    function getMousePos(evt) {
      const rect = ed.canvas.getBoundingClientRect();
      const scaleX = ed.canvas.width / rect.width;
      const scaleY = ed.canvas.height / rect.height;
      return { x: (evt.clientX - rect.left) * scaleX, y: (evt.clientY - rect.top) * scaleY };
    }

    ed.canvas.addEventListener('mousedown', (e) => {
      const pos = getMousePos(e);
      ed.draggingPointIndex = ed.points.findIndex(p => Math.hypot(p.x - pos.x, p.y - pos.y) <= pointRadius * 2);
      if (ed.draggingPointIndex !== -1) ed.draw();
    });

    ed.canvas.addEventListener('mousemove', (e) => {
      if (ed.draggingPointIndex === -1) return;
      const pos = getMousePos(e);
      ed.points[ed.draggingPointIndex].x = pos.x;
      ed.points[ed.draggingPointIndex].y = pos.y;
      ed.draw();
    });

    ed.canvas.addEventListener('mouseup', () => { ed.draggingPointIndex = -1; ed.draw(); });
    ed.canvas.addEventListener('mouseleave', () => { ed.draggingPointIndex = -1; ed.draw(); });
  });

  // Save both
  saveBtn.addEventListener('click', async () => {
    const getNormPoly = (ed) => {
      if (!ed.img.src || ed.points.length === 0) return [];
      return ed.points.map(p => [
        Math.round((p.y / ed.img.height) * 1000),
        Math.round((p.x / ed.img.width) * 1000)
      ]);
    };

    const poly0 = getNormPoly(editors[0]);
    const poly1 = getNormPoly(editors[1]);

    try {
      log("Saving documents... Please wait.");
      const res = await google.colab.kernel.invokeFunction('save_polygon_backend', [poly0, poly1], {});
      const data = res.data && res.data['application/json'];
      if (data && data.logs) {
          data.logs.forEach(l => log(l));
      }
      log("Documentation completely saved.");
    } catch (err) {
      log("Save Error: " + err.message);
    }
  });
</script>
"""

display(HTML(html_code))
````

# **Interactive HuggingFace Adapter Transfer**


````python
# @title Interactive HuggingFace Adapter Transfer
import os
import ipywidgets as widgets
from IPython.display import display, clear_output
from google.colab import drive, userdata
from huggingface_hub import login, HfApi

# 1. Mount Drive & Authenticate
try:
    drive.mount('/content/drive')
except Exception as e:
    print(f"Drive mount issue: {e}")

try:
    # Ensure your secret is named exactly 'HF_TOKEN'
    hf_token = userdata.get('HF_TOKEN')
    login(token=hf_token)
except userdata.SecretNotFoundError:
    print("Error: Could not find 'HF_TOKEN' in Colab Secrets. Please add it and run the cell again.")
    hf_token = None

# 2. UI Components
style = {'description_width': 'initial'}
layout = widgets.Layout(width='100%')

folder_input = widgets.Text(
    value='/content/drive/MyDrive/',
    description='Source Folder Path:',
    style=style, layout=layout,
    placeholder='Enter the absolute path to your checkpoint folder'
)

load_btn = widgets.Button(description="Scan Folder", button_style='info')
repo_input = widgets.Text(
    description='Repo ID (org/name):',
    style=style, layout=layout,
    placeholder='e.g., organization-name/model-name'
)
gated_checkbox = widgets.Checkbox(
    value=True,
    description='Make Repository Gated (Require access requests)',
    style=style
)
upload_btn = widgets.Button(description="Upload Selected Files", button_style='success')
upload_btn.disabled = True # Disabled until files are successfully loaded

file_checkboxes = widgets.VBox()
out = widgets.Output()

# 3. Logic: Scan Folder
def scan_folder(b):
    with out:
        clear_output()
        folder_path = folder_input.value.strip()

        if not os.path.exists(folder_path):
            print(f"Error: Folder not found at {folder_path}.")
            file_checkboxes.children = []
            upload_btn.disabled = True
            return

        files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

        if not files:
            print("Warning: The specified folder is empty or contains no files.")
            file_checkboxes.children = []
            upload_btn.disabled = True
            return

        print(f"Found {len(files)} files. Select the files to upload:")

        # Create checkboxes, auto-checking recommended files for adapter inference
        checkboxes = []
        for f in files:
            is_recommended = any(f.startswith(prefix) for prefix in ['adapter', 'tokenizer', 'chat_', 'processor']) \
                             and not f.endswith('.pt') and not f.endswith('.pth') and not f.endswith('.bin') or f == "adapter_model.bin"

            cb = widgets.Checkbox(value=is_recommended, description=f, style=style, layout=widgets.Layout(width='auto'))
            checkboxes.append(cb)

        file_checkboxes.children = checkboxes
        upload_btn.disabled = False

load_btn.on_click(scan_folder)

# 4. Logic: Upload Files
def upload_files(b):
    with out:
        clear_output()
        if not hf_token:
            print("Error: Cannot proceed because the Hugging Face token is missing.")
            return

        repo_id = repo_input.value.strip()
        if not repo_id or "/" not in repo_id:
            print("Error: Please provide a valid Repo ID format ('organization/model-name').")
            return

        selected_files = [cb.description for cb in file_checkboxes.children if cb.value]
        if not selected_files:
            print("Warning: No files selected for upload.")
            return

        api = HfApi()
        folder_path = folder_input.value.strip()

        try:
            print(f"Verifying repository '{repo_id}' (creating if it does not exist)...")
            api.create_repo(repo_id=repo_id, repo_type="model", exist_ok=True)

            # Handle Gated Status via README metadata
            if gated_checkbox.value:
                print("Applying gated configuration to repository metadata...")
                readme_path = os.path.join(folder_path, "README.md")
                readme_content = ""

                # Check if a local README exists to preserve its content
                if "README.md" in selected_files and os.path.exists(readme_path):
                    with open(readme_path, "r", encoding="utf-8") as f:
                        readme_content = f.read()

                    # Inject gating metadata if missing
                    if "gated:" not in readme_content:
                        if readme_content.startswith("---"):
                            readme_content = readme_content.replace("---\n", "---\ngated: true\n", 1)
                        else:
                            readme_content = "---\ngated: true\n---\n" + readme_content
                else:
                    # Generate a new gated README
                    readme_content = "---\ngated: true\n---\n# Gated Adapter Repository\n\nAccess to this model is restricted. Please request access."
                    selected_files.append("README.md")

                # Upload the modified or generated README
                api.upload_file(
                    path_or_fileobj=readme_content.encode("utf-8"),
                    path_in_repo="README.md",
                    repo_id=repo_id,
                    repo_type="model"
                )

                # Remove README from the selected_files list to prevent duplicate upload attempts
                if "README.md" in selected_files:
                    selected_files.remove("README.md")

            print(f"Starting upload sequence for {len(selected_files)} files...")
            for idx, file_name in enumerate(selected_files, 1):
                file_path = os.path.join(folder_path, file_name)
                print(f"[{idx}/{len(selected_files)}] Uploading {file_name}...")

                api.upload_file(
                    path_or_fileobj=file_path,
                    path_in_repo=file_name,
                    repo_id=repo_id,
                    repo_type="model"
                )
            print(f"Upload complete. You can view your repository here: https://huggingface.co/{repo_id}")

        except Exception as e:
            print(f"Upload failed due to an error: {e}")

upload_btn.on_click(upload_files)

# 5. Display the UI
display(widgets.VBox([
    widgets.HTML("<h3>Hugging Face Adapter Uploader</h3>"),
    folder_input,
    load_btn,
    widgets.HTML("<hr>"),
    file_checkboxes,
    widgets.HTML("<hr>"),
    repo_input,
    gated_checkbox,
    upload_btn,
    out
]))
````
